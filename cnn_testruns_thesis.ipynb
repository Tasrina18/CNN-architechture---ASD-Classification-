CNN ARCHITECHTURE STARTS, FIRST WE COMPUTE OUR INPUT MATRICES (functional connectivity matrix using abide.roiscc400 time series data)

```python
from nilearn.connectome import ConnectivityMeasure


conn_est = ConnectivityMeasure(kind = 'correlation', vectorize=True,discard_diagonal=True)
connect_matrices = conn_est.fit_transform(abide.rois_cc400)
y = abide.phenotypic['DX_GROUP']
y[y==2] = 0

```

PCA applied

```python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import time


start_time = time.time()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(connect_matrices)


n_components = 529
pca = PCA(n_components=n_components)
X_reduced = pca.fit_transform(X_scaled)
end_time = time.time()


elapsed_time = end_time - start_time
print(f"Time elapsed: {elapsed_time} seconds")


explained_variance = np.sum(pca.explained_variance_ratio_)
print(f"Explained Variance with {n_components} components: {explained_variance * 100:.2f}%")

```

SPLTTING OF DATA INTO TRANING, VALIDATION AND TEST SETS

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split (x2, y, test_size = 0.2,random_state=100)
x_test, x_val, y_test, y_val = train_test_split(x_test, y_test, test_size=0.5, random_state=100)
#training = 828 reduced to 670 samples after taking only autism samples
#validation = 104 changed to 84 samples
#testing = 103 changed to 84 samples
#total sample size reduced from 1035 to 838

#changing class weights to deal with class imbalance
count_class_1 = np.sum(y_train == 0)
count_class_2 = np.sum(y_train == 1)
ratio = count_class_1/count_class_2
class_weights={1:ratio, 0:1}
#oversampling using SMOTE to increase class 1 samples
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
original_shape = x_train.shape
x_train = x_train.reshape(x_train.shape[0], -1) # flattens data for each sample into a 1D array
x_train, y_train = smote.fit_resample(x_train, y_train) #oversampling done for the minority class
x_train = x_train.reshape(-1, *original_shape[1:]) # reshaping to original shape done - undoing of the flattening

original_shape = x_test.shape
x_test = x_test.reshape(x_test.shape[0], -1)
x_test, y_test = smote.fit_resample(x_test, y_test)
x_test = x_test.reshape(-1, *original_shape[1:])

original_shape = x_val.shape
x_val = x_val.reshape(x_val.shape[0], -1)
x_val, y_val = smote.fit_resample(x_val, y_val)
x_val = x_val.reshape(-1, *original_shape[1:])

from keras.utils import to_categorical
y_train = to_categorical(y_train, num_classes=2)
y_val = to_categorical(y_val, num_classes=2)
y_test = to_categorical(y_test, num_classes=2)
```

BUILDING THE CONVOLUTIONAL NETWORK


```python
def runmodel():
  import tensorflow as tf
  from tensorflow import keras
  from tensorflow.keras.models import Sequential
  from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten, Dropout
  from sklearn.model_selection import StratifiedKFold
  from sklearn.metrics import classification_report, confusion_matrix
  from sklearn.model_selection import cross_val_score
  import numpy as np
  from keras import regularizers

  # we are using functional API and not sequential API for our cnn code
  S = 23
  # Define the input shape and number of filters
  input_shape = (S, S, 1) # The shape of the connectivity matrix
  num_filters = 32 # The number of filters for each size


  # input layer
  inputs = tf.keras.Input(shape=input_shape)

  # Define a convolution layer with filter size 1 and width 392
  conv1 = tf.keras.layers.Conv2D(filters=num_filters, kernel_size=(1, S), activation='relu', kernel_regularizer=regularizers.l2(0.0001))(inputs)
  pool1 = tf.keras.layers.MaxPool2D(pool_size=(S-1+1, 1))(conv1)
  flat1 = tf.keras.layers.Flatten()(pool1)

  # Define a convolution layer with filter size 2 and width 392
  conv2 = tf.keras.layers.Conv2D(filters=num_filters, kernel_size=(2, S), activation='relu', kernel_regularizer=regularizers.l2(0.0001))(inputs)
  pool2 = tf.keras.layers.MaxPool2D(pool_size=(S-2+1, 1))(conv2)
  flat2 = tf.keras.layers.Flatten()(pool2)

  # Define a convolution layer with filter size 3 and width 392
  conv3 = tf.keras.layers.Conv2D(filters=num_filters, kernel_size=(3, S), activation='relu', kernel_regularizer=regularizers.l2(0.0001))(inputs)
  pool3 = tf.keras.layers.MaxPool2D(pool_size=(S-3+1, 1))(conv3)
  flat3 = tf.keras.layers.Flatten()(pool3)

  # Define a convolution layer with filter size 4 and width 392
  conv4 = tf.keras.layers.Conv2D(filters=num_filters, kernel_size=(4, S), activation='relu', kernel_regularizer=regularizers.l2(0.0001))(inputs)
  pool4 = tf.keras.layers.MaxPool2D(pool_size=(S-4+1, 1))(conv4)
  flat4 = tf.keras.layers.Flatten()(pool4)

  # Define a convolution layer with filter size 5 and width 392
  conv5 = tf.keras.layers.Conv2D(filters=num_filters, kernel_size=(5, S), activation='relu', kernel_regularizer=regularizers.l2(0.0001))(inputs)
  pool5 = tf.keras.layers.MaxPool2D(pool_size=(S-5+1, 1))(conv5)
  flat5 = tf.keras.layers.Flatten()(pool5)

  # Define a convolution layer with filter size 6 and width 392
  conv6 = tf.keras.layers.Conv2D(filters=num_filters, kernel_size=(6, S), activation='relu', kernel_regularizer=regularizers.l2(0.0001))(inputs)
  pool6 = tf.keras.layers.MaxPool2D(pool_size=(S-6+1, 1))(conv6)
  flat6 = tf.keras.layers.Flatten()(pool6)

  # Define a convolution layer with filter size 7 and width 392
  conv7 = tf.keras.layers.Conv2D(filters=num_filters, kernel_size=(7, S), activation='relu', kernel_regularizer=regularizers.l2(0.0001))(inputs)
  pool7 = tf.keras.layers.MaxPool2D(pool_size=(S-7+1, 1))(conv7)
  flat7 = tf.keras.layers.Flatten()(pool7)

  # Concatenate the outputs of all convolution layers
  concat = tf.keras.layers.Concatenate()([flat1, flat2, flat3,flat4, flat5, flat6,flat7])

  # Apply dropout with rate 0.25
  drop = tf.keras.layers.Dropout(0.3)(concat)

  #Define a fully connected layer with 64 units and tanh activation
  #dense = tf.keras.layers.Dense(64, activation='tanh')(drop)

  # Define the output layer with one unit and sigmoid activation for binary classification
  outputs = tf.keras.layers.Dense(2, activation='softmax')(drop)

  # Create the model
  model = tf.keras.Model(inputs=inputs, outputs=outputs)

  # Compile the model with binary crossentropy loss and Adam optimizer
  model.compile(loss='categorical_crossentropy', optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), metrics=['accuracy'])
  model.summary()
  return model







```

```python
model = runmodel()
```

FITTING OUR DATA INTO THE CNN MODEL

```python
import time


start_time = time.time()
batch_size = 32
epochs = 30
hist = model.fit(x_train, y_train, batch_size=batch_size,epochs=epochs,class_weight=class_weights,validation_data=(x_val, y_val),verbose=1)
end_time = time.time()

elapsed_time = end_time - start_time
print(f"Time elapsed: {elapsed_time} seconds")
```

For plotting Confusion Matrix

```python
import pandas as pd
import numpy as np
import seaborn as sns

def prep_submissions(preds_array, file_name='abc.csv'):
    preds_df = pd.DataFrame(preds_array)
    predicted_labels = preds_df.idxmax(axis=1) #convert back one hot encoding to categorical variabless
    return predicted_labels

def draw_confusion_matrix(true,preds, figsize=(6, 6)):
    conf_matx = confusion_matrix(true, preds)
    fig, ax = plt.subplots(figsize=figsize)
    disp = ConfusionMatrixDisplay(confusion_matrix=conf_matx, display_labels=np.unique(true))

    disp.plot(cmap="Blues", values_format='g',ax=ax)
    plt.show()
```

Model evaluation

```python
score = model.evaluate(x_test, y_test, verbose = 1)
y_preds = model.predict(x_test)
test_preds_labels = prep_submissions(y_preds)
y_test_labels = np.argmax(y_test, axis=1) # this is done to convert y_test back to 1D array for printing confusion matrix
draw_confusion_matrix(y_test_labels, test_preds_labels)
print('Test loss:', score[0])
print('Test accuracy:', score[1])
```
